```python
# Muestra qué categorías tienen registros sin la columna "artist"
# y cuántos faltan en cada una.

category_artist_status = (
    df_grammys
    .groupby("category",dropna=False)
    .agg(
        total_rows=("year", "size"),
        artist_nulos=("artist",lambda s: s.isna().sum()),
        artist_no_nulos=("artist", lambda s: s.notna().sum())
    )
    .reset_index()
    .assign(
        artist_coverage_pct=lambda x: (x["artist_no_nulos"] / x["total_rows"] * 100).round(2)
    )
    .sort_values(["artist_nulos", "artist_coverage_pct"], ascending=[False, True])
)

display(
    category_artist_status[category_artist_status["artist_nulos"] > 0]
)
```

```python
# Categorías donde artist está informado en todos los registros
categorias_artist_obligatorio = (
    category_artist_status
    .loc[
        (category_artist_status["total_rows"] > 0)
        & (category_artist_status["artist_nulos"] == 0),
        "category"
    ]
    .tolist()
)

# Categorías con cobertura parcial: requieren revisión
categorias_artist_revision = (
    category_artist_status
    .loc[
        (category_artist_status["artist_nulos"] > 0)
        & (category_artist_status["artist_no_nulos"] > 0),
        ["category", "total_rows", "artist_nulos", "artist_coverage_pct"]
    ]
    .sort_values("artist_coverage_pct")
)

print(f"Categorías con artist siempre informado: {len(categorias_artist_obligatorio)}")
print(f"Categorías con cobertura parcial: {len(categorias_artist_revision)}")

display(categorias_artist_revision)
```
